```python
for num in range(1, 51):
    if num > 1:
        for i in range(2, num):
            if num % i == 0:
                break
        else:
            print(num, end=" ")
```

```python
# Explanation
# 1. range(1, 51) – Generates numbers from 1 to 50.
# 2. if num > 1 – Excludes 1 because it is not a prime number.
# 3. for i in range(2, num) – Checks divisibility from 2 to num-1.
# 4. num % i == 0 – Checks whether the number is divisible by another number.
# 5. break – Stops the loop if a divisor is found.
# 6. else – Executes when the loop finishes without break, meaning the number is prime.
# 7. print(num, end=" ") – Prints the prime numbers on the same line.
```
