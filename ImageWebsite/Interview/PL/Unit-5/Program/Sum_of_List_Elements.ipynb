```python
numbers = []

n = int(input("Enter number of elements: "))

for i in range(n):
    num = int(input("Enter a number: "))
    numbers.append(num)

total = 0

for num in numbers:
    total = total + num

print("List:", numbers)
print("Sum of list elements:", total)
```

```python
# Explanation
# 1. numbers = [] – Creates an empty list.
# 2. n = int(input()) – Takes the number of elements from the user.
# 3. for i in range(n) – Repeats the loop to accept elements.
# 4. numbers.append(num) – Adds each entered number to the list.
# 5. total = 0 – Initializes the sum variable.
# 6. for num in numbers – Iterates through each list element.
# 7. total = total + num – Adds each element to the total.
# 8. print() – Displays the list and its sum.
```
