```python
import csv

threshold = 80

file = open("students.csv", "r")

reader = csv.DictReader(file)

print("Students scoring above", threshold, ":")

for row in reader:
    if int(row["Marks"]) > threshold:
        print(row)

file.close()
```

```python
# Explanation
# 1. import csv – Imports the CSV module.
# 2. threshold = 80 – Sets the minimum marks threshold.
# 3. open("students.csv", "r") – Opens the CSV file in read mode.
# 4. csv.DictReader(file) – Reads each row as a dictionary using column names as keys.
# 5. for row in reader: – Iterates through all rows.
# 6. int(row["Marks"]) > threshold – Checks whether marks exceed 80.
# 7. print(row) – Displays only the rows satisfying the condition.
# 8. file.close() – Closes the file.
```

```python
# Difference
# csv.reader()								csv.DictReader()
# Returns rows as lists.					Returns rows as dictionaries.
# Access data using index.					Access data using column name.
# Example: row[2]							Example: row["Marks"]
# Header is also read as a normal row.		Header is used as dictionary keys.
```
