```python
import csv

file = open("students.csv", "w", newline="")

writer = csv.writer(file)

writer.writerow(["Name", "Age", "Marks"])
writer.writerow(["Ankit", 20, 85])
writer.writerow(["Rahul", 21, 90])
writer.writerow(["Aman", 20, 75])
writer.writerow(["Priya", 21, 95])
writer.writerow(["Neha", 20, 60])

file.close()

print("CSV file created successfully.")
```
