```python
my_tuple = ()

n = int(input("Enter number of elements: "))

elements = []

for i in range(n):
    value = input("Enter an element: ")
    elements.append(value)

my_tuple = tuple(elements)

reverse_tuple = my_tuple[::-1]

print("Original Tuple:", my_tuple)
print("Reversed Tuple:", reverse_tuple)
```

```python
# Explanation
# 1. my_tuple = () – Initializes an empty tuple.
# 2. n = int(input()) – Takes the number of elements from the user.
# 3. elements = [] – Creates an empty list to store the entered elements.
# 4. for i in range(n): – Repeats the loop according to the number of elements.
# 5. value = input() – Accepts an element from the user.
# 6. elements.append(value) – Adds each element to the list.
# 7. my_tuple = tuple(elements) – Converts the list into a tuple.
# 8. my_tuple[::-1] – Reverses the tuple using slicing.
# 9. print() – Displays the original and reversed tuples.
```
