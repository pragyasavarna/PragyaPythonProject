```python
import random
import matplotlib.pyplot as plt

# Simulate 5 temperature readings
readings = []

for i in range(5):
    temperature = random.randint(20, 40)
    readings.append(temperature)

print("Temperature Readings:", readings)

# Plot the temperature graph
plt.plot([1, 2, 3, 4, 5], readings, marker='o')

plt.title("IoT Temperature Monitoring")
plt.xlabel("Reading Number")
plt.ylabel("Temperature (°C)")

plt.show()
```
