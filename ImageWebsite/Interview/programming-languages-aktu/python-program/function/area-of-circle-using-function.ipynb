```python
def calculate_area(radius):
    area = 3.14 * radius * radius
    return area

radius = float(input("Enter the radius of circle: "))

result = calculate_area(radius)

print("Area of circle:", result)
```

```python
# Explanation
# 1. def calculate_area(radius): – Defines a function that accepts the radius.
# 2. area = 3.14 * radius * radius – Calculates the area of the circle.
# 3. return area – Returns the calculated area.
# 4. input() – Takes the radius from the user.
# 5. calculate_area(radius) – Calls the function and passes the radius.
# 6. print() – Displays the calculated area.
```
