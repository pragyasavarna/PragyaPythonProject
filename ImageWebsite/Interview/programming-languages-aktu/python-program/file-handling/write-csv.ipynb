```python
import csv

file = open("students.csv", "w", newline="")

writer = csv.writer(file)

writer.writerow(["Name", "Age", "Marks"])
writer.writerow(["Ankit", 20, 85])
writer.writerow(["Rahul", 21, 90])
writer.writerow(["Aman", 20, 75])
writer.writerow(["Priya", 21, 95])
writer.writerow(["Neha", 20, 60])

file.close()

print("CSV file created successfully.")
```

```python
# Explanation
# 1. import csv – Imports the CSV module, which provides functions to read and write CSV files.
# 2. file = open("students.csv", "w", newline="") – Creates or opens the students.csv file in write mode ("w"). If the file already exists, its previous data is overwritten.
# 3. newline="" – Prevents extra blank lines from appearing between rows when writing data into a CSV file, especially on Windows.
# 4. writer = csv.writer(file) – Creates a CSV writer object that is used to write data into the CSV file.
# 5. writer.writerow() – Writes a single row of data into the CSV file. It is used to write column headings and student records.
# 6. file.close() – Closes the file after writing all the records and saves the changes.
# 7. print("CSV file created successfully.") – Displays a success message on the screen.
```
