```python
file = open("student.txt", "r")

n = int(input("Enter number of last lines to read: "))

lines = file.readlines()

print("Last", n, "lines of the file:")

for line in lines[-n:]:
    print(line, end="")

file.close()
```

```python
# Explanation:
# - open() opens the file in read mode ("r").
# - readlines() reads all lines and stores them in a list.
# - lines[-n:] selects the last n lines from the list.
# - The for loop displays the selected lines.
# - close() closes the file.
# In Python, the minus (-) sign is used because negative indexing selects elements from the end of a list.
# Understanding lines[-n:]
# Suppose n = 3.
# lines[-3:]
# - -3 means start from the third element from the end.
# - : means select all elements from that position up to the end.
# - Therefore, it returns the last 3 lines.
```
