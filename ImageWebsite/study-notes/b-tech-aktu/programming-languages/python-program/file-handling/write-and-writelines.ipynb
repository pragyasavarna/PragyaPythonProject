```python
f = open("student.txt", "w")

# Using write()
f.write("Name: Ankit Gupta\n")
f.write("Course: B.Tech\n")

# Using writelines()
lines = ["Subject: Python\n", "Year: First Year\n"]
f.writelines(lines)

f.close()

print("Data written successfully")
```

```python
# 1. write() Function
# The write() function is used to write a single string into a file.
# It writes the given text at the current position of the file.
# It does not automatically add a newline character.
# Syntax:
# file.write(string)
# 2. writelines() Function
# The writelines() function is used to write multiple strings into a file.
# It accepts a list or other iterable containing strings.
# It does not automatically insert new lines between them.
# Syntax:
# file.writelines(list)
```
