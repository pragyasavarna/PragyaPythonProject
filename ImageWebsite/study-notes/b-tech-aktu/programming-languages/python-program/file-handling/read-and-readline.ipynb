```python
# Writing data into a file
f = open("student.txt", "w")
f.write("Name: Ankit\n")
f.write("Course: B.Tech\n")
f.write("Subject: Python\n")
f.close()

# Using read() function
f = open("student.txt", "r")
print("Using read():")
print(f.read())
f.close()

# Using readline() function
f = open("student.txt", "r")
print("Using readline:")
print(f.readline())
print("Using readline with end:")
print(f.readline(), end="")
print(f.readline(), end="")
f.close()
```

```python
# The read() function is used to read the entire content of a file or a specified number of characters from it.
# Syntax:
# file.read()
# The readline() function is used to read one line at a time from a file. Each time it is called, it reads the next line.
# Syntax:
# file.readline()
```

```python
# **Explanation of Program**

# 1. `open("student.txt", "w")` – Creates a file named `student.txt` in write mode.
# 2. `f.write()` – Writes three lines of student information into the file. `\n` moves the text to the next line.
# 3. `f.close()` – Closes the file after writing.
# 4. `open("student.txt", "r")` – Opens the file in read mode.
# 5. `f.read()` – Reads and displays the complete content of the file.
# 6. `f.readline()` – Reads one line at a time. Here, it is called twice to read the first two lines.
# 7. `f.close()` – Closes the file after reading.
```
