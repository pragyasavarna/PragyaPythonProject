```python
set1 = {10, 20, 30, 40, 50}
set2 = {30, 40, 60, 70}

result = set1 - set2

print("Difference of two sets:", result)
```

```python
# Explanation:
# - set1 and set2 are two sets containing different elements.
# - set1 - set2 returns elements that are present in set1 but not in set2.
# - The result is stored in the result variable and displayed using print().
```
