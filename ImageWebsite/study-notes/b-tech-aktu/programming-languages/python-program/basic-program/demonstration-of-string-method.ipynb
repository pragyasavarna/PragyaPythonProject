```python
text = "Machine Learning "

print("Original String:", text)

print("Lowercase:", text.lower())

print("Uppercase:", text.upper())

print("After strip():", text.strip())

print("After replace():", text.replace("Machine", "Deep"))

print("Find Learning:", text.find("Learning"))

print("Count of a:", text.count("a"))
```

```python
# Explanation:
# - lower() converts the string into lowercase letters.
# - upper() converts the string into uppercase letters.
# - strip() removes extra spaces from the beginning and end.
# - replace() replaces "Machine" with "Deep".
# - find() returns the starting index of "Learning".
# - count() counts how many times the letter "a" appears in the string.
```
