```python
# The pass statement is a null statement. It does nothing when executed.
# It is used as a placeholder when a statement is required syntactically but no action is needed.
```

```python
for i in range(1, 6):
    if i == 3:
        pass
    print(i)
```

```python
# The main use of pass is to keep a block of code empty temporarily without causing an error.
# In Python, if you write an if, for, while, or function definition, its body cannot be empty.
# You can use pass as a placeholder and add the actual code later.
# Example 1: Using pass in an if statement
```

```python
age = 20

if age >= 18:
    pass
else:
    print("Not eligible")
```
