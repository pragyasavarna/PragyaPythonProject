```python
str1 = "Python Programming"

print("Original String:", str1)
print("First 6 characters:", str1[0:6])
print("Characters from index 7 to 17:", str1[7:18])
print("First 6 characters:", str1[:6])
print("Last 11 characters:", str1[-11:])
print("Reverse String:", str1[::-1])
```

```python
# Explanation:
# - str1[0:6] – Extracts characters from index 0 to 5.
# - str1[7:18] – Extracts characters from index 7 to 17.
# - str1[:6] – Extracts the first 6 characters.
# - str1[-11:] – Extracts the last 11 characters using negative indexing.
# - str1[::-1] – Reverses the entire string.
```
