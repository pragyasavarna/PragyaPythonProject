```python
numbers = [10, 20, 30, 40, 50, 60, 70, 80]
```

```python
print("Original List:", numbers)
print("First 4 elements:", numbers[0:4])
print("Elements from index 2 to 5:", numbers[2:6])
print("First 3 elements:", numbers[:3])
print("Last 3 elements:", numbers[-3:])
print("Reverse List:", numbers[::-1])
print("Alternate elements:", numbers[::2])
```

```python
# Explanation:
# - numbers[0:4] – Returns elements from index 0 to 3.
# - numbers[2:6] – Returns elements from index 2 to 5.
# - numbers[:3] – Returns the first 3 elements.
# - numbers[-3:] – Returns the last 3 elements.
# - numbers[::-1] – Reverses the list.
# - numbers[::2] – Returns alternate elements.
```
